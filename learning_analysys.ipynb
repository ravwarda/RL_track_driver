```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the training metrics CSV file
metrics_file = "training_metrics.csv"
data = pd.read_csv(metrics_file)

# Plot Episode Rewards
plt.figure(figsize=(10, 6))
plt.plot(data["Episode"], data["Reward"], label="Episode Reward", color="blue")
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.title("Rewards Over Time")
plt.legend()
plt.grid()
plt.show()

# Calculate Moving Average Reward
window_size = 10
data["Moving Average Reward"] = data["Reward"].rolling(window=window_size).mean()

# Plot Moving Average Reward
plt.figure(figsize=(10, 6))
plt.plot(data["Episode"], data["Moving Average Reward"], label=f"Moving Average Reward (window={window_size})", color="orange")
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.title("Moving Average Reward Over Time")
plt.legend()
plt.grid()
plt.show()

# Plot Actor and Critic Loss
plt.figure(figsize=(10, 6))
plt.plot(data["Episode"], data["Actor Loss"], label="Actor Loss", color="red")
plt.ylim(0, 10)
plt.xlabel("Episode")
plt.ylabel("Loss")
plt.title("Actor Loss Over Time")
plt.legend()
plt.grid()
plt.show()

# Plot Actor and Critic Loss
plt.figure(figsize=(10, 6))
plt.plot(data["Episode"], data["Critic Loss"], label="Critic Loss", color="green")
plt.ylim(0, 50)
plt.xlabel("Episode")
plt.ylabel("Loss")
plt.title("Critic Loss Over Time")
plt.legend()
plt.grid()
plt.show()
```
