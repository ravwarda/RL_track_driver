```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the training metrics CSV file
metrics_file = "training_metrics.csv"
data = pd.read_csv(metrics_file)

# Plot Episode Rewards
plt.figure(figsize=(10, 6))
plt.plot(data["Step"], data["Reward"], label="Episode Reward", color="blue")
plt.xlabel("Step")
plt.ylabel("Reward")
plt.title("Rewards Over Time")
plt.legend()
plt.grid()
plt.show()

# Plot Actor and Critic Loss
plt.figure(figsize=(10, 6))
plt.plot(data["Step"], data["Actor Loss"], label="Actor Loss", color="red")
plt.xlabel("Step")
plt.ylabel("Loss")
plt.title("Actor Loss Over Time")
plt.legend()
plt.grid()
plt.show()

# Plot Actor and Critic Loss
plt.figure(figsize=(10, 6))
plt.plot(data["Step"], data["Critic Loss"], label="Critic Loss", color="green")
plt.xlabel("Step")
plt.ylabel("Loss")
plt.title("Critic Loss Over Time")
plt.legend()
plt.grid()
plt.show()
```
